```python
import camelot

pdf_path = r"data\00007.pdf"

# Extract tables from page 1 using 'lattice' flavor (default)
tables = camelot.read_pdf(pdf_path, pages="1", flavor="lattice")

print(f"Total tables found: {tables.n}")

# Iterate through extracted tables
for i, table in enumerate(tables):
    print(f"\n--- Table {i + 1} ---")
    print(f"Parsing accuracy: {table.parsing_report['accuracy']:.2f}%")
    
    # Convert to a Pandas DataFrame
    df = table.df
    # print(df.head())
    
    # # Export options:
    # # table.to_csv(f"table_{i+1}.csv")
    # # table.to_excel(f"table_{i+1}.xlsx")
    # # table.to_json(f"table_{i+1}.json")
```

```python
import camelot
import pandas as pd

pdf_path = r"data/00007.pdf"

# Extract tables across all pages
tables = camelot.read_pdf(pdf_path, pages="all", flavor="lattice")

print(f"Total tables found across entire PDF: {tables.n}")

# 1. Inspect each extracted table
for i, table in enumerate(tables):
    print(f"\n--- Table {i + 1} (Page {table.page}) ---")
    print(f"Accuracy: {table.parsing_report['accuracy']:.2f}%")
    display(table.df)  # In Jupyter notebook use display(), or print(table.df)
```

```python
# Combine all extracted tables
all_dfs = [table.df for table in tables]
combined_df = pd.concat(all_dfs, ignore_index=True)

# Save to Excel
combined_df.to_excel("extracted_full.xlsx", index=False)
```
